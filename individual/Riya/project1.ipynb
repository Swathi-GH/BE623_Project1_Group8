Q2 (a)

```python
with open("selenok.fasta", "r") as fh:
   lines = fh.readlines() #to open the fasta file
mrna = ""     

for line in lines:
    if not line.startswith(">"):     
        mrna = mrna + line.replace("\n", "")    #to remove any spaces in the sequence and include only bases
    
start_base = 72           #starts the sequence at 72 
end_base = 356            #ends the sequence at 356

cds = mrna[start_base-1:end_base]   #subtracting 1 because indexing starts from 0
print(cds) 
```

```python
print(len(cds))   #to print the length of the nucleotides
```

```python
print(len(cds)/3)  #to confirm if it is divisible by 3
```

Q2 (b)

```python
genetic_code = {
    #Phenylalanine (F)
    "TTT": "F", "TTC": "F",
    #Leucine (L)
    "TTA": "L", "TTG": "L", "CTT": "L",  
    "CTC": "L", "CTA": "L", "CTG": "L",
    #Isoleucine (I)
    "ATT": "I", "ATC": "I", "ATA": "I",
    #Methionine (M) which is a start codon
    "ATG": "M",
    #Valine (V)
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    #Serine (S)
    "TCT": "S", "TCC": "S", "TCA": "S",
    "TCG": "S", "AGT": "S", "AGC": "S",
    #Proline (P)
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    #Threonine (T)
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    #Alanine (A)
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    #Tyrosine (Y)
    "TAT": "Y", "TAC": "Y",
    #Histidine (H)
    "CAT": "H", "CAC": "H",
    #Glutamine (Q)
    "CAA": "Q", "CAG": "Q",
    #Asparagine (N)
    "AAT": "N", "AAC": "N",
    #Lysine (K)
    "AAA": "K", "AAG": "K",
    #Aspartic acid (D)
    "GAT": "D", "GAC": "D",
    #Glutamic acid (E)
    "GAA": "E", "GAG": "E",
    #Cysteine (C)
    "TGT": "C", "TGC": "C",
    #Tryptophan (W)
    "TGG": "W",
    #Arginine (R)
    "CGT": "R", "CGC": "R", "CGA": "R", 
    "CGG": "R", "AGA": "R", "AGG": "R",
    #Glycine (G)
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
    #Stop codons
    "TAA": "*", "TAG": "*", "TGA": "*",     #for selenocysteine TGA codes for U
}

prtn = ""     #Translating the sequence into protein

for i in range(0, len(cds), 3):      
    codon = cds[i:i+3]               #to read one codon at a time 
    aa = genetic_code[codon]         #to extract amino acid from the dictionary
    prtn += aa                       #to add amino acids to protein sequence
    
print(prtn[:30])                     #to print the first 30 amino acids
print(len(prtn))                     #to print the length of the protein 
```

```python
print(prtn)      #to print the whole protein sequence
```

Q2 (c)

```python
with open("selenok_prtn.fasta", "r") as fh: 
    lines = fh.readlines()   #to read all file lines
prtn_seq = ""        #to store the protein sequence
for line in lines:
    if not line.startswith(">"):   #to remove headers in the file
        prtn_seq = prtn_seq + line.replace("\n", "")   #to remove spaces in the file
print(prtn_seq)
```

```python
p1 = prtn
p2 = prtn_seq
differences = 0     #to count the number of differences
for x,y in zip(p1,p2):
    if x != y:      #to check if there are any differences
        differences = differences + 1
print(differences)
```
