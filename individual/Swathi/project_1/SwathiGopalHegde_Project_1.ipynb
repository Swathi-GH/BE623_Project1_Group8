Q2 (a): Extracting the CDS region

```python
with open("TP53_sequence.fasta", "r") as sequence:   #opening TP53 mrna fasta file in read mode
    seq = sequence.read() #saving it as variable 'seq'
print(seq) #printing the sequence
```

```python
print(len(seq)) #printing the length of the sequence
```

```python
start =  143 #start coordinate of CDS
end = 1324 #end coordinate of CDS
CDS_region = seq[start-1:end] #extracting the CDS region; since index begins with 0, sequence starts from 142nd position
print(CDS_region) #printing CDS region
```

```python
print(len(CDS_region)) #length of CDS region
```

```python
print(len(CDS_region)/3) #checking if it is divisible by 3, for codon correctness
```

Q2(b): Building the genetic code 

```python
genetic_code = {
    "TTT" : "F",
    "TTC" : "F",
    "TTA" : "L",
    "TTG" : "L",
    "CTT" : "L",
    "CTC" : "L",
    "CTA" : "L",
    "CTG" : "L",
    "ATT" : "I",
    "ATC" : "I",
    "ATA" : "I",
    "ATG" : "M",
    "GTT" : "V",
    "GTC" : "V",
    "GTA" : "V",
    "GTG" : "V",
    "TCT" : "S",
    "TCC" : "S",
    "TCA" : "S",
    "TCG" : "S",
    "CCT" : "P",
    "CCC" : "P",
    "CCA" : "P",
    "CCG" : "P",
    "ACT" : "T",
    "ACC" : "T",
    "ACA" : "T",
    "ACG" : "T",
    "GCT" : "A",
    "GCC" : "A",
    "GCA" : "A",
    "GCG" : "A",
    "TAT" : "Y",
    "TAC" : "Y",
    "TAA" : "*",
    "TAG" : "*",
    "CAT" : "H",
    "CAC" : "H",
    "CAA" : "Q",
    "CAG" : "Q",
    "AAT" : "N",
    "AAC" : "N",
    "AAA" : "K",
    "AAG" : "K",
    "GAT" : "D",
    "GAC" : "D",
    "GAA" : "E",
    "GAG" : "E",
    "TGT" : "C",
    "TGC" : "C",
    "TGA" : "*",
    "TGG" : "W",
    "CGT" : "R",
    "CGC" : "R",
    "CGA" : "R",
    "CGG" : "R",
    "AGT" : "S",
    "AGC" : "S",
    "AGA" : "R",
    "AGG" : "R",
    "GGT" : "G",
    "GGC" : "G",
    "GGA" : "G",
    "GGG" : "G",
} #genetic code for the 64 standard amino acids
```

```python
protein = "" #defining empty string for the protein sequence to be translated
for i in range(0,len(CDS_region),3): #for loop for iterating through the entire mrna sequence
    codon = CDS_region[i:i+3] #codon includes 3 nucleotides
    amino_acid = genetic_code[codon] #extracting the genetic code from the created list
    protein = protein + amino_acid #concatenation to make the protein sequence
    
print(protein[0:30]) #first 30 amino acids of protein sequence
print(len(protein)) #length of complete protein sequence
```

```python
print(protein) #prints protein sequence 
```

Q2(c): Comparing both the sequences

```python
with open("TP53_protein_new.fasta", "r") as p: #open protein sequence downloaded from ncbi
    protein_ncbi = p.read()
print(protein_ncbi) #print protein sequence
```

```python
#comparing custom translated protein sequence and protein sequence from ncbi
s1 = protein #custom sequence
s2 = protein_ncbi #sequence from ncbi
differences = 0 #setting differences to 0
for a,b in zip(s1,s2):
    if a != b: #if position of s1 is not equal to s2
        differences = differences + 1
print(differences)
```

```python
from Bio.Seq import Seq

s = Seq(CDS_region)
s = s.translate() #translating the sequence 
print(s)
```

```python
print (s == protein) #to check if seq translated by biopython is equal to custom translated sequence
```
