```python
from Bio import SeqIO

```

```python
record = SeqIO.read("nd5.fasta", "fasta")
print(len(record.seq))
start_cds= 1
end_cds= 1812
length = record.seq[start_cds-1:end_cds]
print(len(length))
#dividing by 3 to get number of codons
print(len(length)/3)


```

```python

#creating dictionary for mitochondrial genetic code
genetic_code = {
    "TTT": "F",
    "TTC": "F",
    "TTA": "L",
    "TTG": "L",
    "CTT": "L",
    "CTC": "L",
    "CTA": "L",
    "CTG": "L",
    "ATT": "I",
    "ATC": "I",
    "ATA": "M",
    "ATG": "M",
    "GTT": "V",
    "GTC": "V",
    "GTA": "V",
    "GTG": "V",
    "TCT": "S",
    "TCC": "S",
    "TCA": "S",
    "TCG": "S",
    "CCT": "P",
    "CCC": "P",
    "CCA": "P",
    "CCG": "P",
    "ACT": "T",
    "ACC": "T",
    "ACA": "T",
    "ACG": "T",
    "GCT": "A",
    "GCG": "A",
    "GCA": "A",
    "GCC": "A",
    "TAT": "Y",
    "TAC": "Y",
    "TAA": "*",
    "TAG": "*",
    "CAT": "H",
    "CAC": "H",
    "CAA": "Q",
    "CAG": "Q",
    "AAT": "N",
    "AAC": "N",
    "AAA": "K",
    "AAG": "K",
    "GAT": "D",
    "GAC": "D",
    "GAA": "E",
    "GAG": "E",
    "TGT": "C",
    "TGC": "C",
    "TGA": "W",
    "TGG": "W",
    "CGT": "R",
    "CGC": "R",
    "CGA": "R",
    "CGG": "R",
    "AGT": "S",
    "AGC": "S",
    "AGA": "*",
    "AGG": "*",
    "GGT": "G",
    "GGC": "G",
    "GGA": "G",
    "GGG": "G"

}  
    
    
protein = ""
for i in range(0,len(length),3):
    codon = str(length[i:i+3])
    protein += genetic_code[codon]

print(protein)
               
    
```

```python
print("First 30 residues:", protein[:30])
print("Total length:", len(protein))
```

```python
with open("nd5_protein.fasta", "r") as fh:
    lines = fh.readlines()

deposited_seq =""
for l in lines:
     if not l.startswith(">"):
        deposited_seq += l.strip()    
```

```python
file1 = deposited_seq
file2 = protein
differences = 0
for a, b in zip(file1, file2):
    if a != b:
        differences = differences + 1
print(differences)
print(len(deposited_seq))
                         
        
```
